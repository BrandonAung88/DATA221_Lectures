```python
def average(list1 = [1]):
    sum_of_list1 = 0
    for i in list1:
        sum_of_list1 = sum_of_list1 + i
    length_of_list1 = len(list1)

    average_of_list1 = sum_of_list1 / length_of_list1
    print(average_of_list1)

average([1])
```

```python
# Function with return output in Python

def power(x = 1, y = 1):
    result = x ** y
    return(result)

square = power(x = 3, y = 2)
print(square)
```

```python
def division_fucntion (var_x = 1, var_y =1):
    if var_y == 0:
        return("Invalid :Zero Division!")
    else:
        return(var_x / var_y)

a = division_fucntion()
print(a)

b = division_fucntion(var_x = 1, var_y = 0)
print(b)
```
