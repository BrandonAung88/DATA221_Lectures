```python
#stdin stands for user input

# Text files can be read, written, and extended using open function

# "r": for reading
# "w" for writing a new file
# "a" for appending to the end of the file

file = open("Readme.txt","r")
print(file.read())
file.close()
```

```python
file = open("Readme.txt","a")
print(file.write("Hello I added This"))
file.close()
```

```python
from bs4 import BeautifulSoup
import requests

uofc_grading_system_html = requests.get("https://data.who.int/dashboards/covid19/cases?n=c").text
parsed_html_document = BeautifulSoup(uofc_grading_system_html, "html5lib")

all_paragraph_elements = parsed_html_document.find_all("p") #finds every HTML paragraph element (<p>) 

first_paragraph_element = all_paragraph_elements[0] 
print(first_paragraph_element.text)
```
