```python
# Nested If-Else Loops
final_marks_of_data_221 = 95

if final_marks_of_data_221 >= 99:
    print("Letter Grade: 'A+'")
elif 99>final_marks_of_data_221>=95:
    print("Letter Grade: 'A'")
elif 95>final_marks_of_data_221>=90:
    print("Letter Grade: 'A-'")
elif 90>final_marks_of_data_221>=85:
    print("Letter Grade: 'B+'")
elif 85>final_marks_of_data_221>=80:
    print("Letter Grade: 'B'")
elif 80>final_marks_of_data_221>=75:
    print("Letter Grade: 'B-'")
elif 75>final_marks_of_data_221>=70:
    print("Letter Grade: 'C+'")
elif 70>final_marks_of_data_221>=65:
    print("Letter Grade: 'C'")
elif 65>final_marks_of_data_221>=60:
    print("Letter Grade: 'C-'")
```

```python
# For Loop
s = 0
for i in range(1,3):
    s = s + i

print(s)


x = 0
for y in [1,2,3]:
    x = x + y

print(x)
```

```python
# For loop in one line

l = [i + 1 for i in range(2,5)]  #this makes a list
print(l)
```

```python
#Exercise

list_of_number_grades = [90,99,88]
list_of_letter_grades = []

for final_marks_of_data_221 in list_of_number_grades:
    if final_marks_of_data_221 >= 99:
        list_of_letter_grades.append("A+")
    elif 99>final_marks_of_data_221>=95:
        list_of_letter_grades.append("A-")
    elif 95>final_marks_of_data_221>=90:
        list_of_letter_grades.append("A")
    elif 90>final_marks_of_data_221>=85:
        list_of_letter_grades.append("B+")
    elif 85>final_marks_of_data_221>=80:
        list_of_letter_grades.append("B")
    elif 80>final_marks_of_data_221>=75:
        list_of_letter_grades.append("B-")
    elif 75>final_marks_of_data_221>=70:
        list_of_letter_grades.append("C+")
    elif 70>final_marks_of_data_221>=65:
        list_of_letter_grades.append("C")
    elif 65>final_marks_of_data_221>=60:
        list_of_letter_grades.append("C-")

print(list_of_letter_grades)
```
