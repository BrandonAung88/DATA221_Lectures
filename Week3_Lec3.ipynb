```python
# Calculating mean of the 

from sklearn.preprocessing import StandardScaler
import numpy as np  # Import NumPy for mean calculation

data = np.array([[10, 90], [12, 96], [9, 120], [13, 112], [8, 88]])  # Convert to NumPy array
mean_data = np.mean(data, axis=0)


print(f"Original means per feature: {mean_data}" )  # [10.4 101.2][code_file:1]

scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
mean_scaled_data = np.mean(scaled_data, axis=0)

print(f"Scaled means per feature: {mean_scaled_data}")  # ≈ [0. 0.][code_file:1]
print("Scaler computed means:", scaler.mean_)  # Exactly [10.4 101.2][code_file:1][web:94]
```
